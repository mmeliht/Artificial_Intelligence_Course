# ANN -  DEEP Learning - Concrete Compressive Strength - Regression  & Classification Project

# Beton Basınç Dayanımı Tahmini

Bu proje, betonun **basınç dayanımını** karışımındaki malzemeler ve kürlenme süresine göre tahmin etmeyi ve betonun yapısal sınıfını sınıflandırmayı amaçlamaktadır. Veri seti [UCI Machine Learning Repository](https://archive.ics.uci.edu/ml/datasets/Concrete+Compressive+Strength) kaynağından alınmıştır.

## Amaçlar
1. **Regresyon:** Çimento, Slag, Uçucu Kül (FlyAsh), Su, Plastiklaştırıcı, Grov ve İnce Agrega ile Betonun yaşına (Age) bağlı olarak basınç dayanımını (MPa) tahmin etmek.
2. **Sınıflandırma:** Aynı özellikleri kullanarak betonun sınıfını tahmin etmek (non-structural, residential, commercial, high-strength).

## Yaklaşım
- Veri ön işleme ve özellik-hedef ayrımı.
- Eğitim-test veri seti bölünmesi ve ölçekleme.
- **klasik makine öğrenmesi algoritmaları** ve **Derin Öğrenme (TensorFlow/Keras)** kullanılarak regresyon ve sınıflandırma modelleri geliştirme.
- Model performansını regresyon için R² ve MSE, sınıflandırma için doğruluk (accuracy) ile değerlendirme.

Bu temel model, beton dayanımı tahmini ve sınıflandırması için performans kriterlerini belirlemektedir.
