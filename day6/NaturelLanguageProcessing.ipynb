```python
text = ['Call you tonight', 'call me a cab', 'please call me ...  PLEASE']
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
vect = CountVectorizer()
```

```python
vect.fit_transform(text).toarray()
```

```python
import pandas as pd
```

```python
tf = pd.DataFrame(vect.fit_transform(text).toarray() , columns = vect.get_feature_names_out())
```

```python
tf
```

```python
df = vect.fit_transform(text).toarray().sum(axis = 0)
```

```python
df
```

```python
tf/df
```

```python
df = pd.read_csv('yelp.csv')
```

```python
df.head()
```

```python
df['text'][0]
```

```python
df['text']=df['text'].str.lower() # küçük harf yapar
df['text']=df['text'].str.replace('[^\w\s]','',regex=True) # noktalama kaldır
df['text']=df['text'].str.replace('\d+','',regex=True) # rakamları kaldır
df['text']=df['text'].str.replace('\n','',regex=True) # satır sonlarını kaldır
df['text']=df['text'].str.replace('\r','',regex=True) # enterleri kaldır
```
