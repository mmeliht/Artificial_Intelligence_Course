## Feature Engineering

### Screen Resolution

```python
df['ScreenResolution'].value_counts()
```

```python
df["Resolution"] = df["ScreenResolution"].str.extract(r"(\d{3,4}x\d{3,4})", expand=False) ## ekran çözünürlüğü ayrıldı
```

```python
df["Resolution"].value_counts()
```

```python
df["Touch"] = df["ScreenResolution"].str.contains("touchscreen", case=False, na=False).astype(int) ## dokunmatik ekran ayrıldı
```

```python
df["Touch"].value_counts()
```

```python
df["Panel"] = np.where(df["ScreenResolution"].str.contains(r"ips|retina", case=False, na=False), "IPS", pd.NA)
```

```python
df["Panel"].isna().sum()
```

### Screen Resolution End
