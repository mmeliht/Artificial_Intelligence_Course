### Memory

```python
df['Memory'].value_counts()
```

```python
import re

_pat = re.compile(
    r'(?P<val>\d+(?:\.\d+)?)\s*(?P<unit>TB|GB)\s*(?P<typ>SSD|HDD|Hybrid|Flash(?:\s+Storage)?)',
    re.IGNORECASE
)

def _parse_memory(s: str):
    out = {'SSD_GB': 0.0, 'HDD_GB': 0.0, 'Flash_GB': 0.0, 'Hybrid_GB': 0.0, 'Disk_Count': 0}
    if not isinstance(s, str):
        out['Total_GB'] = 0.0
        return out

    for m in _pat.finditer(s):
        val = float(m.group('val'))
        unit = m.group('unit').upper()
        typ  = m.group('typ').strip().lower()

        # Flash Storage sadeleştir
        if typ.startswith('flash'):
            typ_norm = 'Flash'
        elif typ == 'ssd':
            typ_norm = 'SSD'
        elif typ == 'hdd':
            typ_norm = 'HDD'
        else:
            typ_norm = 'Hybrid'

        gb = val * 1024.0 if unit == 'TB' else val
        out[f'{typ_norm}_GB'] += gb
        out['Disk_Count'] += 1

    out['Total_GB'] = out['SSD_GB'] + out['HDD_GB'] + out['Flash_GB'] + out['Hybrid_GB']
    return out

# Özellikleri çıkar
feats = df['Memory'].apply(_parse_memory).apply(pd.Series)

# Bayraklar ve oranlar
for t in ['SSD', 'HDD', 'Flash', 'Hybrid']:
    feats[f'Has_{t}'] = (feats[f'{t}_GB'] > 0).astype(int)
    feats[f'{t}_Ratio'] = (feats[f'{t}_GB'] / feats['Total_GB']).replace([np.inf, -np.inf], 0).fillna(0)

# İstersen kapasite sütunlarını tam sayıya yuvarla
for col in ['SSD_GB', 'HDD_GB', 'Flash_GB', 'Hybrid_GB', 'Total_GB']:
    feats[col] = feats[col].round().astype(int)

# Son DataFrame
df = pd.concat([df, feats], axis=1)

# Kontrol için ilk satırlar
print(df.head())
## ChatGpt fonksiyonu
```

```python
def parse_cpu(cpu_str):
    parts = {}
    
    # Marka
    if "Intel" in cpu_str:
        parts["CpuBrand"] = "Intel"
    elif "AMD" in cpu_str:
        parts["CpuBrand"] = "AMD"
    else:
        parts["CpuBrand"] = "Other"
    
    # Aile (Core i3/i5/i7/i9, Ryzen, Pentium, Celeron, Atom, Xeon)
    match_family = re.search(r"(Core i[3579]|Ryzen \d|Pentium|Celeron|Atom|Xeon)", cpu_str)
    parts["CpuFamily"] = match_family.group(0) if match_family else "Other"
    
    # Model kodu (örn: 7200U, 8550U, 9420)
    match_model = re.search(r"\d{4,}[A-Z]?", cpu_str)
    parts["CpuModel"] = match_model.group(0) if match_model else "Unknown"
    
    # Saat hızı (GHz → float)
    match_speed = re.search(r"(\d\.\d+)GHz", cpu_str)
    parts["CpuSpeedGHz"] = float(match_speed.group(1)) if match_speed else None
    
    return pd.Series(parts)
```

```python
cpu_features = df["Cpu"].apply(parse_cpu)
```

```python
# Orijinal tabloya ekle
df = pd.concat([df, cpu_features], axis=1)
```

### GPU

```python
def parse_gpu(gpu_str: str):
    s = str(gpu_str)
    parts = {}

    # --- Marka (Brand) ---
    if re.search(r"\b(nvidia|geforce|quadro)\b", s, re.I):
        parts["GpuBrand"] = "Nvidia"
    elif re.search(r"\b(amd|radeon)\b", s, re.I):
        parts["GpuBrand"] = "AMD"
    elif re.search(r"\bintel\b", s, re.I):
        parts["GpuBrand"] = "Intel"
    elif re.search(r"\b(apple|m1|m2|m3)\b", s, re.I):
        parts["GpuBrand"] = "Apple"
    elif re.search(r"\b(arm|mali|adreno|powerVR)\b", s, re.I):
        parts["GpuBrand"] = "ARM"
    else:
        parts["GpuBrand"] = "Other"

    # --- Aile / Sınıf (Family/Class) ---
    family = "Other"
    family_patterns = [
        (r"\brtx\b", "RTX"),
        (r"\bgtx\b", "GTX"),
        (r"\bquadro\b", "Quadro"),
        (r"\bgeforce\b", "GeForce"),
        (r"\bradeon\s*rx\b", "Radeon RX"),
        (r"\bradeon\s*pro\b", "Radeon Pro"),
        (r"\bvega\b", "Vega"),
        (r"\bradeon\b", "Radeon"),
        (r"\biris\s*plus\b", "Iris Plus"),
        (r"\biris\b", "Iris"),
        (r"\buhd\b", "UHD Graphics"),
        (r"\bhd\s*graphics\b", "HD Graphics"),
        (r"\barc\b", "Intel Arc"),
    ]
    for pat, lab in family_patterns:
        if re.search(pat, s, re.I):
            family = lab
            break
    parts["GpuFamily"] = family

    # --- Model (sayısal/sonek) ---
    model = None
    model_patterns = [
        r"RTX\s*(\d{3,4}\s*(?:Ti|SUPER)?)",
        r"GTX\s*(\d{3,4}\s*(?:Ti|M|MX)?)",
        r"Quadro\s*([A-Z]?\d{3,4})",
        r"GeForce\s*(?:[A-Z]*\s*)?(\d{3,4}\s*(?:Ti|MX)?)",
        r"Radeon\s*RX\s*(\d{3,4}\s*(?:XT)?)",
        r"Radeon\s*Pro\s*(\d{3,4})",
        r"Radeon\s*(HD\s*\d{4})",
        r"Vega\s*(\d{1,2})",
        r"Iris\s*Plus\s*Graphics\s*(\d{3,4})",
        r"UHD\s*Graphics\s*(\d{3})",
        r"HD\s*Graphics\s*(\d{3,4})",
        r"Arc\s*(A\d{3})",
    ]
    for mp in model_patterns:
        m = re.search(mp, s, re.I)
        if m:
            model = m.group(1).strip()
            break
    if not model:
        m_fallback = re.search(r"\b(\d{3,4})(?:[A-Z]{0,2})\b", s)
        if m_fallback:
            model = m_fallback.group(1)
    parts["GpuModel"] = model if model else "Unknown"

    # --- Entegre mi? (Integrated) ---
    s_low = s.lower()
    integrated = False
    if parts["GpuBrand"] == "Intel":
        integrated = True
    elif "integrated" in s_low:
        integrated = True
    elif parts["GpuBrand"] == "AMD":
        if re.search(r"radeon\s*(rx|pro|vega)", s_low):
            integrated = False
        elif re.search(r"radeon\s*(r\d|graphics)", s_low):
            integrated = True
        else:
            integrated = False
    elif parts["GpuBrand"] == "Nvidia":
        integrated = False
    else:
        integrated = False

    parts["GpuIntegrated"] = integrated
    return pd.Series(parts)

# Yeni GPU sütunlarını üret ve birleştir
gpu_features = df["Gpu"].apply(parse_gpu)
df = pd.concat([df, gpu_features], axis=1)

# Kontrol için ilk 20 satır
print(df[["Gpu", "GpuBrand", "GpuFamily", "GpuModel", "GpuIntegrated"]].head(20))
```
