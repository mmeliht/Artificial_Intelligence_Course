```python
import pandas as pd
```

```python
df = pd.read_csv('restaurant.csv')
```

```python
df.head()
```

```python
df['text']
```

```python
df['text']=df['text'].str.lower() # küçük harf yapar
df['text']=df['text'].str.replace('[^\w\s]','',regex=True) # noktalama kaldır
df['text']=df['text'].str.replace('\d+','',regex=True) # rakamları kaldır
df['text']=df['text'].str.replace('\n','',regex=True) # satır sonlarını kaldır
df['text']=df['text'].str.replace('\r','',regex=True) # enterleri kaldır
```

```python
df['text']
```

```python
# Sentimel Analysis
```

```python
ybw=df[(df.stars==1)|(df.stars==5)]
```

```python
ybw
```

```python
ybw.reset_index(drop=True,inplace=True)
```

```python
ybw
```

```python
x=ybw[['text']]
y=ybw[['stars']]
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from textblob import TextBlob
```

```python
def ekkok(text):
    words = TextBlob(text).words
    return [word.lemmatize() for word in words  ]
```

```python
vect = CountVectorizer(analyzer = ekkok, stop_words = 'english', ngram_range = (1,2))
```

```python
xyeni = vect.fit_transform(x['text'])
```

```python
print(vect.get_feature_names_out()[-50:])
```

```python
x_train , x_test , y_train , y_test = train_test_split(xyeni , y , random_state= 42 , test_size = .20)
```

```python
dt = DecisionTreeClassifier()
```

```python
model = dt.fit(x_train , y_train)
```

```python
tahmin = dt.predict(x_test)
```

```python
accuracy_score(y_test , tahmin)
```
