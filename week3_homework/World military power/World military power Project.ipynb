```python
import pandas as pd
pd.set_option("display.max_columns",100)

import warnings
warnings.filterwarnings('ignore')

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
### Sütun isimleri birinci satırdan itibaren başlıyor . o yüzden sütun isimlerini birinci satırdan itibaren almalıyız
```

```python
df_raw = pd.read_excel('World military power.xlsx')
```

```python
df = df_raw[2:].copy()
df.columns = df_raw.iloc[0]
df.reset_index(drop=True, inplace=True)
```

```python
df_country = df.iloc[:, 0]        # İlk ülke adını al
df_values = df.iloc[:, 1::2]      # Tek indeksli sayısal sütunları al

df = pd.concat([df_country, df_values], axis=1)
```

```python
# Boş değerler doğrudan bir ülkeyle ilgili bilgi verdiği için veri oldurma yöntemleri yanlış bir yaklaşım olacağını düşünüyorum
# o yüzden nan değerleri 0 olarak düzelteceğim.
```

```python
df = df.fillna(0)
```

```python
df.info()
```

```python
df.sample(10)
```

```python
# tüm sütunları int yapmak için
# virgülle ayrılmış sayılar olduğu için önce string yapıp virgülleri kaldırdık. sonra int e çevirdik
for col in df.columns[2:]:
    df[col] = df[col].astype(str).str.replace(",", "").astype(int)
```

## VISUALIZATION

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Stil ayarları
sns.set_theme(style="whitegrid")
plt.rcParams["font.size"] = 12
```

```python
top_tanks = df[[df.columns[0], "Tank Strength value"]].sort_values(by="Tank Strength value", ascending=False).head(10)

plt.figure(figsize=(12, 7))
sns.barplot(data=top_tanks, y=df.columns[0], x="Tank Strength value", palette="Blues_r")
plt.title("En Çok Tanka Sahip 10 Ülke", fontsize=16, weight='bold')
plt.xlabel("Tank Sayısı")
plt.ylabel("Ülke")
plt.tight_layout()
plt.show()
```

```python
top_budget = df[[df.columns[0], "defense spending budget value"]].sort_values(by="defense spending budget value", ascending=False).head(10)

plt.figure(figsize=(12, 7))
colors = sns.color_palette("Reds", len(top_budget))
sns.barplot(data=top_budget, y=df.columns[0], x="defense spending budget value", palette=colors)
plt.title("En Yüksek Savunma Bütçesine Sahip 10 Ülke", fontsize=16, weight='bold')
plt.xlabel("Savunma Bütçesi (USD)")
plt.ylabel("Ülke")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
scatter = sns.scatterplot(
    data=df,
    x="Tank Strength value",
    y="defense spending budget value",
    hue=df.columns[0],
    palette="tab20",
    s=100,
    legend=False,
    alpha=0.7
)

# En yüksek 5 ülkeyi isimlendir. Tüm ülke isimleri yazınca çürkin bir görüntü oluşuyor
top5 = df.sort_values(by="defense spending budget value", ascending=False).head(5)
for _, row in top5.iterrows():
    plt.text(row["Tank Strength value"], row["defense spending budget value"], row[df.columns[0]], fontsize=10)

plt.title("📈 Savunma Bütçesi ve Tank Sayısı İlişkisi", fontsize=16, weight='bold')
plt.xlabel("Tank Sayısı")
plt.ylabel("Savunma Bütçesi")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Corelasyon grafiği
# Sadece sayısal sütunları al
df_numeric = df.select_dtypes(include=['int64', 'float64'])

# Korelasyon matrisi
corr = df_numeric.corr()

plt.figure(figsize=(14, 12))
sns.heatmap(corr, cmap="coolwarm", annot=False, fmt=".2f", linewidths=0.5)
plt.title("Askeri Değişkenler Arası Korelasyon Haritası", fontsize=16, weight='bold')
plt.tight_layout()
plt.show()
```

```python
top5 = df.sort_values(by="defense spending budget value", ascending=False).head(5)
components = ["Tank Strength value", "Aircraft Strength value", "Submarine Fleet Strength value"]

top5_comp = top5[[df.columns[0]] + components].set_index(df.columns[0])

top5_comp.plot(kind="bar", stacked=True, figsize=(12,7), colormap="tab20")
plt.title(" En Güçlü 5 Ülkenin Askeri Bileşenleri (Stacked Bar)", fontsize=14)
plt.xlabel("Ülke")
plt.ylabel("Toplam Varlık")
plt.tight_layout()
plt.show()
```

```python
# Find Elbow Point
```

```python
from yellowbrick.cluster import KElbowVisualizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
km=KMeans()
```

```python
x = df.select_dtypes(include=["int64", "float64"])
```

```python
vis=KElbowVisualizer(km,k=(2,10))
vis.fit(x)
vis.show()
```

```python
# Model
```

```python
model = KMeans(4)
```

```python
model = model.fit(x)
```

```python
tahmin = model.predict(x)
```

```python
tahmin
```

```python
x['cluster'] = tahmin
```

```python
x.sample(15)
```

```python
pd.value_counts(x['cluster'])
```

```python
silhouette_score(x , tahmin)
```

```python

# Görselleştirme
plt.figure(figsize=(10,6))
plt.scatter(x["Tank Strength value"], x["defense spending budget value"], c=tahmin, cmap='viridis', s=100, alpha=0.8)
plt.colorbar(label="Küme No")
plt.xlabel("Tank Sayısı")
plt.ylabel("Savunma Bütçesi (USD)")
plt.title("K-Means Kümeleme: Tank vs Savunma Bütçesi", fontsize=14)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import DBSCAN # baska bir clustering algoritmasi
```

```python
db = DBSCAN(min_samples= 5)
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import StandardScaler
x2=StandardScaler().fit_transform(x)
y=db.fit_predict(x2)

plt.scatter(x['Tank Strength value'],x['defense spending budget value'],c=y,cmap='autumn')
```

```python
from sklearn.cluster import AgglomerativeClustering  ##

agg_clustering = AgglomerativeClustering(n_clusters=4)
y_agg = agg_clustering.fit_predict(x)

# Plot the results
plt.scatter(x., x.Price, c=y_agg, s=50, cmap='viridis')
plt.title("Agglomerative Clustering")
plt.show()
```
