```python
def mis_val_cal(df) :
    mis_val = df.isnull().sum() 
    mis_val_percent = 100 * df.isnull().sum() / len(df) # eksik verilerin yüzde kaç olduğunu bulduk
    mis_val_table = pd.concat([mis_val, mis_val_percent], axis=1) # eksik verilerin sayısıyla yüzdesini birleştirdik
    mis_val_table_ren_columns = mis_val_table.rename(
        columns = {0 : 'Missing Values', 1 : '% of Total Values'}) # kolon ismi verdik
    return mis_val_table_ren_columns
```

```python
mis_val_cal(df)
```

```python
df = df.drop(df[['Months since last delinquent']] , axis = 1) # kolonun yüzde 50 den fazlası boş olduğu için kolonu kaldırdık
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
early_stop = EarlyStopping() # epoch sayısı ne kadar olursa olsun en başarılı epoch sayısını bulunca durur
```

```python
model = Sequential()
model.add(Dense(120, activation='relu'))
model.add(Dense(256, activation='relu'))
model.add(Dense(128, activation='relu'))
model.add(Dense(3, activation='relu'))  
model.add(Dense(4, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = model.fit(df_x, df_y, validation_split=0.2, epochs=130, batch_size=32, callbacks=[early_stop], verbose=1)
```
