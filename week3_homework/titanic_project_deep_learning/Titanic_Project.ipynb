### Normalization and Deep Learning

```python
train = df[:891]
test = df[891:]
```

```python
test['Survived'].head()
```

```python
x = train.drop('Survived','PassengerId', axis = 1)
y = train[['Survived']]
```

```python
#pip install tensorflow
```

```python
import tensorflow as tf
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import Dense
```

```python
from sklearn.preprocessing import normalize, scale
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
x = StandardScaler().fit_transform(x)
```

```python
model = Sequential()
model.add(Dense(120, activation='relu'))
model.add(Dense(256, activation='relu'))
model.add(Dense(128, activation='relu'))
model.add(Dense(3, activation='relu'))  
model.add(Dense(4, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = model.fit(x ,y , epochs = 300, batch_size = 32 , validation_split = 0.10 ,verbose = 1)
```

```python
x_test = test.drop(['Survived'], axis=1)
y_test = test[['Survived']]
```

```python
# Test verisi ile modelin doğruluğunu değerlendirme
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print(f'Test Kaybı: {test_loss}, Test Doğruluğu: {test_accuracy}')
```

```python
# Yeni veriler üzerinde tahmin yapma
predictions = model.predict(x_test)

# Tahmin sonuçlarını gösterme
print(predictions[:10])  # İlk 10 tahmini göster
```
