```python
# eğitim seviyelerini toparla
df["education"] = df["education"].replace({
    "college/university": "student",
    "two-year": "student",
    "masters": "student",
    "ph.d": "graduated",
    "law": "graduated",
    "med": "graduated",
    "high": "graduated",
    "space": "student"
}, regex=True)
```
