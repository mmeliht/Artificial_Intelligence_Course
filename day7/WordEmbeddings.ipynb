```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red','green','blue','red','green'],
    'size': ['small','medium','large','medium','small'],
    'price': [10,15,20,8,12]
})
```

```python
df
```

```python
le = LabelEncoder()
```

```python
df['size2'] = le.fit_transform(df['size'] ) ## bedenleri sıralayabileceğimiz için le kullandık. 0 dan 2 ye
```

```python
df
```

```python
oe = OneHotEncoder()
```

```python
oe.fit_transform(df[['color']]).toarray() # renklerde üstünlük olmadığı için bunu kullandık
```
