```python
from qiskit import QuantumCircuit # Kuantum Devreleri
qc = QuantumCircuit(2,2)
qc.h(0)
qc.measure(0,0)
qc.draw('mpl')
```

```python
from qiskit import QuantumCircuit # Kuantum Devreleri
qc = QuantumCircuit(2,2)
qc.h(0)
qc.cx(0,1)
qc.measure([0,1],[0,1])
qc.draw('mpl')
```

```python
from qiskit import QuantumCircuit # Kuantum Devreleri
qc = QuantumCircuit(3,3)
qc.h(1)
qc.cx(1,2)
qc.cx(0,1)
qc.h(0)
qc.measure([0,1],[0,1])
qc.cx(1,2)
qc.cz(0,2)
qc.measure(2,2)
qc.draw('mpl')
```

```python
from qiskit import QuantumCircuit # Kuantum Devreleri
qc = QuantumCircuit(1,1)
qc.x(0)
qc.measure(0,0)
qc.draw('mpl')
```

```python
from qiskit import QuantumCircuit # Kuantum Devreleri
qc = QuantumCircuit(2,2)
qc.h([0,1])
qc.measure([0,1],[0,1])
qc.draw('mpl')
```

```python
from qiskit import QuantumCircuit # Kuantum Devreleri
qc = QuantumCircuit(3,3)
qc.h(0)
qc.cx(0,1)
qc.cx(0,1)

qc.measure([0,1,2],[0,1,2])
qc.draw('mpl')
```

```python
from qiskit import QuantumCircuit # Kuantum Devreleri
qc = QuantumCircuit(3,1)
qc.h([0,1,2])
qc.cz(0,2)
qc.cz(1,2)
qc.h([0,1])
qc.measure(0,0)
qc.draw('mpl')
```
